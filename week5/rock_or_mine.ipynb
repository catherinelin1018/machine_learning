```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("../datasets/ROCK_OR_MINE.csv")
```

```python
df["R"] = df["R"].replace({"R": 0, "M": 1})
```

```python
df["R"]
```

```python
X = df.drop("R", axis=1)
y = df["R"]
```

```python
X_train, X_test, y_train, y_test =  train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
```

```python
rand_forest_model = RandomForestClassifier()

rand_forest_model.fit(X_train, y_train)
rand_pred = rand_forest_model.predict(X_test)
```

```python
print(confusion_matrix(y_test, rand_pred))
print("---------------------------------")
print(classification_report(y_test, rand_pred))
```

```python
print(f"model score on training data: {rand_forest_model.score(X_train, y_train)}")
print(f"model score on testing data: {rand_forest_model.score(X_test, y_test)}")
```

```python
rand_forest_model = RandomForestClassifier()

rand_forest_model.fit(X_train, y_train)
rand_pred = rand_forest_model.predict(X_test)
```

```python
rand_forest_model_entropy = RandomForestClassifier(criterion="entropy", random_state=42)

rand_forest_model_entropy.fit(X_train, y_train)
rand_pred_entropy = rand_forest_model_entropy.predict(X_test)

print(f"model score on training data: {rand_forest_model_entropy.score(X_train, y_train)}")
print(f"model score on testing data: {rand_forest_model_entropy.score(X_test, y_test)}")
print("---------------------------------")
print(confusion_matrix(y_test, rand_pred_entropy))
print("---------------------------------")
print(classification_report(y_test, rand_pred_entropy))
```

```python
rand_forest_model_log = RandomForestClassifier(criterion="log_loss", random_state=42)

rand_forest_model_log.fit(X_train, y_train)
rand_pred_log = rand_forest_model_log.predict(X_test)

print(f"model score on training data: {rand_forest_model_log.score(X_train, y_train)}")
print(f"model score on testing data: {rand_forest_model_log.score(X_test, y_test)}")
print("---------------------------------")
print(confusion_matrix(y_test, rand_pred_log))
print("---------------------------------")
print(classification_report(y_test, rand_pred_log))
```
