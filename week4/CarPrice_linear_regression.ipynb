```python
df['drivewheel'].unique()
```

```python
car_type = pd.get_dummies(df['carbody'], drop_first=True)
drive_wheel = pd.get_dummies(df['drivewheel'], drop_first=True)
fuel_system = pd.get_dummies(df['fuelsystem'], drop_first=True)

df = pd.concat([df, car_type, drive_wheel, fuel_system], axis=1)
```

```python
df.info()
```

```python
df.drop(['carbody', 'drivewheel', 'fuelsystem'], axis=1, inplace=True)
```

```python
df.info()
```

```python
engine_type = pd.get_dummies(df['enginetype'], drop_first=True)
df = pd.concat([df, engine_type], axis=1)
```

```python
df.head()
```

```python
df.drop(['enginetype', 'CarName', 'car_ID'], axis=1, inplace=True)
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop("price", axis=1),
                                                    df["price"],
                                                    test_size = 0.2,
                                                    random_state=101
                                                   )
```

```python
linear_model = LinearRegression()
```

```python
linear_model.fit(X_train, y_train)
```

```python
predictions = linear_model.predict(X_test)
```

```python
from sklearn import metrics
```

```python
print('MAE: ', metrics.mean_absolute_error(y_test, predictions))
print('MSE: ', metrics.mean_squared_error(y_test, predictions))
print('RMSE: ', np.sqrt(metrics.mean_squared_error(y_test, predictions)))
```
